```python
a = [1, 2, 3]
b = [4, 5]
zipped = zip(a, b)

print(list(zipped))
```

```python
a = [[1, 2], [3, 4], [5, 6]]
b = [4, 5, 6]
c = [7, 8, 9]
zipped = zip(a, b, c)

print(list(zipped))
```

```python
a = [[1, 2], [3, 4], [5, 6]]
zipped = zip(*a)  # 꾸러미를 푼다

print(list(zipped))
```

```python
a = [[1, 2], [3, 4], [5, 6]]
zipped = zip(*a[::-1])  # 꾸러미를 푼다

print(list(zipped))
```

```python
a = [0, 1, 0]
b = [1, 1, 1]

zipped = zip(*(a, b)[::-1])

print(list(zipped))
```

```python
a = [1, 1, 0]
b = [2, 1, 1]

zipped = list(zip(*(a, b)))[::-1]

print(zipped)
```

```python
a = [1, 1, 0]
b = [2, 1, 1]

zipped = list(zip(*(a, b)))

print(zipped)
```
